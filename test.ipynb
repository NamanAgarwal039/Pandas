```python
import pandas as pd

data ={
    "Name":['naman','aman','Ram','shyam','amit','ankit','arun','rohit','mohit','krish'],
    "Age":[23,22,32,19,20,23,22,26,24,25],
    "Salary":[56000,340000,None,43000,41000,44000,45000,42000,39000,40000],
    "Performence_Score":[80,None,92,87,85,88,84,83,84,88]
}

df = pd.DataFrame(data)

print(df)
print("\n")

#By Using drop 

# df.dropna(inplace=True)
# print(df)


#by using fill any particular data
df.fillna(0,inplace=True)
print(df)


# Fill string column
df["Name"] = df["Name"].fillna("Unknown")

# Fill numeric columns
df["Salary"] = df["Salary"].fillna(0)
#df["Performance_Score"] = df["Performance_Score"].fillna(0)
df['Performence_Score']=df['Performence_Score'].fillna(0)
print(df)
```

# Interpolation

```python
import pandas as pd
data = {
    'Time':[1,2,3,4,5],
    'value':[10,20,None,30,50]
}

df=pd.DataFrame(data)

print("Before Interpolation")
print(df)

print("\nAfter Interpolation")
df["value"]=df["value"].interpolate(method="linear")
print(df)
```

# Sorting of data

```python
import pandas as pd


data ={
    "Name":['Naman','Aman','Rohan'],
    "Age":[23,22,29],
    'Salary':[20000,39000,15000]
}

df=pd.DataFrame(data)


print("Sorting Age in  ascending order ")
df.sort_values(by='Age',ascending=True,inplace=True)

print(df)
```

# groupby function

```python
import pandas as pd
data ={
    "Name":['Naman','Aman','Rohan'],
    "Age":[23,22,29],
    'Salary':[20000,39000,15000]
}

df=pd.DataFrame(data)
#single column
grouped = df.groupby("Age")["Salary"].sum()

print(grouped)

#multiple group

group = df.groupby(['Age',"Name"])["Salary"].sum()
print(group)
```

# Merging the data


```python

''''
types of joins
1. Inner
2.outer
3.cross
4.right
5.left
'''

import pandas as pd

df_customer = pd.DataFrame({
    'Customer_id':[1,2,3],
    'Customer_name':["Ram","Shyam","Aman"]
})

df_order = pd.DataFrame({
    'Customer_id':[1,2,4],
    'Amount':[450,499,900]
})


#merge
df_merge= pd.merge(df_customer,df_order,on="Customer_id",how="inner")
print("Inner join")
print(df_merge)
```
